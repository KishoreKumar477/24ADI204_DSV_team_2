```python
import os
import kagglehub
import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler

# 1. Dataset import as in DSV project.ipynb
path = kagglehub.dataset_download("blastchar/telco-customer-churn")
data = pd.read_csv(os.path.join(path, "WA_Fn-UseC_-Telco-Customer-Churn.csv"))

# Handle empty spaces and drop customerID exactly as in the notebook
l1 = [len(i.split()) for i in data["TotalCharges"]]
l2 = [i for i in range(len(l1)) if l1[i] != 1]
for i in l2:
    data.loc[i, "TotalCharges"] = data.loc[(i - 1), "TotalCharges"]
data["TotalCharges"] = data["TotalCharges"].astype(float)
data.drop(columns=["customerID"], inplace=True)

# 2. Categorical Encoding (LabelEncoder)
df = data.copy(deep=True)
le = LabelEncoder()
text_data_features = [
    i for i in list(df.columns) if i not in list(df.describe().columns)
]
for i in text_data_features:
    df[i] = le.fit_transform(df[i])

print("--- Data after Categorical Encoding ---")
display(df.head())

# 3. Fixing Skewness (Log Transformation)
numerical_features = ["tenure", "MonthlyCharges", "TotalCharges"]
print("\n--- Skewness Before Transformation ---")
display(df[numerical_features].skew().to_frame(name="Skewness_Before"))

for col in numerical_features:
    df[col] = np.log1p(df[col])

print("\n--- Skewness After Log Transformation ---")
display(df[numerical_features].skew().to_frame(name="Skewness_After"))

# 4. Normalization (MinMaxScaler)
mms = MinMaxScaler()
df_normalized = df.copy()
df_normalized[numerical_features] = mms.fit_transform(
    df_normalized[numerical_features]
)

print("\n--- Data after Normalization (MinMaxScaler) ---")
display(df_normalized[numerical_features].head())

# 5. Scaling (StandardScaler)
ss = StandardScaler()
df_standardized = df.copy()
df_standardized[numerical_features] = ss.fit_transform(
    df_standardized[numerical_features]
)

print("\n--- Data after Scaling (StandardScaler) ---")
display(df_standardized[numerical_features].head())
```
